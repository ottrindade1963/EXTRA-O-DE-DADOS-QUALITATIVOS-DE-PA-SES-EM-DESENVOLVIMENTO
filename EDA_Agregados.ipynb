```python
# 3. Visualizar gráficos gerados (Método 1)
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

pasta = 'eda_metodo1_inner'
graficos = sorted([f for f in os.listdir(pasta) if f.endswith('.png')])

print(f"\n{'═' * 60}")
print(f"  GRÁFICOS — MÉTODO 1 (Inner Join)")
print(f"{'═' * 60}")

for g in graficos:
    fig, ax = plt.subplots(figsize=(12, 6))
    img = mpimg.imread(os.path.join(pasta, g))
    ax.imshow(img)
    ax.axis('off')
    ax.set_title(g.replace('.png', '').replace('_', ' ').title(), fontsize=11)
    plt.show()
```
