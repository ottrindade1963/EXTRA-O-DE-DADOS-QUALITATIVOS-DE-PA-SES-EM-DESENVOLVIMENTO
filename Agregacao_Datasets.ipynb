1. **Inner Join** — Junção interna estrita (Allison, 2001; Wooldridge, 2010)
2. **Left Join + Imputação** — Preservação com interpolação temporal (Schafer & Graham, 2002)
3. **Outer Join + Fonte** — Junção completa rastreável (Rubin, 1976; Graham, 2009)
